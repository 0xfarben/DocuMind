```python
## Create a simple text file

import os

os.makedirs("../data/text_files", exist_ok=True)
```

```python
# TextLoader, there are many loaders in langchain

from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
document = loader.load()
print(document)
```

```python
### Load multiple similar(txt, pdf, etc..) files in a directory

from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader(
    "../data/text_files",
    glob= "**/*.txt",
    loader_cls= TextLoader,
    loader_kwargs= { 'encoding': 'utf-8' },
    show_progress= True
)

# To log the progress of DirectoryLoader you need to install tqdm, `pip install tqdm`

multiple_documents = dir_loader.load()
multiple_documents
```

```python
### Load multiple similar(txt, pdf, etc..) files in a directory

from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

pdf_loader = DirectoryLoader(
    "../data/pdf",
    glob= "**/*.pdf",
    loader_cls= PyMuPDFLoader, # use any one loader. mupdf is better version of pyPdf
    show_progress= True
)

multiple_pdf_documents = pdf_loader.load()
multiple_pdf_documents
```
